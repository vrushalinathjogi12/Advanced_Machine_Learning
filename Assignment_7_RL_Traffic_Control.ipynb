AIM: To implement a Reinforcement Learning model to control traffic signals and reduce waiting time.

ALGORITHM:
1. Define traffic states (number of vehicles)
2. Define actions (Green for Road A / Road B)
3. Initialize Q-table
4. Select action (explore/exploit)
5. Apply action (change signal)
6. Give reward (less waiting = positive reward)
7. Update Q-values
8. Repeat for multiple iterations
9. Select best action for traffic control

```python
import numpy as np
import random

# Define states (traffic levels: low=0, medium=1, high=2)
states = [0, 1, 2]

# Actions: 0 = Green for Road A, 1 = Green for Road B
actions = [0, 1]

# Q-table (state: (A_traffic, B_traffic))
Q = {}

# Learning parameters
alpha = 0.1
gamma = 0.9
epsilon = 0.2

# Get random traffic state
def get_state():
    return (random.choice(states), random.choice(states))

# Reward function
def get_reward(state, action):
    A, B = state

    if action == 0:  # Green for A
        return B - A   # reduce waiting on A
    else:             # Green for B
        return A - B

# Choose action
def choose_action(state):
    if random.random() < epsilon:
        return random.choice(actions)
    else:
        q_values = [Q.get((state, a), 0) for a in actions]
        return actions[np.argmax(q_values)]

# Update Q-value
def update_q(state, action, reward, next_state):
    old_q = Q.get((state, action), 0)
    future_q = max([Q.get((next_state, a), 0) for a in actions])
    Q[(state, action)] = old_q + alpha * (reward + gamma * future_q - old_q)

# Training
for episode in range(500):
    state = get_state()

    for step in range(10):
        action = choose_action(state)
        reward = get_reward(state, action)
        next_state = get_state()

        update_q(state, action, reward, next_state)
        state = next_state

# Testing
print("Traffic Light Control Decisions:")

for _ in range(5):
    state = get_state()
    action = choose_action(state)

    print(f"State (A,B): {state} -> Action:",
          "Green A" if action == 0 else "Green B")
```
