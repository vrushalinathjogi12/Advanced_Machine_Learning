AIM: To implement a machine learning model to classify tweets as positive or negative sentiment.

Algorithm (Simple Steps)
1. Load tweet dataset
2. Clean text (remove symbols, stopwords)
3. Convert text into numerical form (TF-IDF)
4. Split data into training and testing
5. Train model (Naive Bayes)
6. Predict sentiment
7. Evaluate accuracy

```python
# Import libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Sample dataset
data = {
    'tweet': [
        "I love this product",
        "This is the worst experience",
        "Amazing service",
        "I hate this",
        "Very good quality",
        "Not happy with the service"
    ],
    'sentiment': [1, 0, 1, 0, 1, 0]  # 1=Positive, 0=Negative
}

df = pd.DataFrame(data)

# Features and labels
X = df['tweet']
y = df['sentiment']

# Convert text to numerical
vectorizer = TfidfVectorizer()
X_vectorized = vectorizer.fit_transform(X)

# Split data
X_train, X_test, y_train, y_test = train_test_split(X_vectorized, y, test_size=0.2)

# Model
model = MultinomialNB()
model.fit(X_train, y_train)

# Predict
predictions = model.predict(X_test)

# Accuracy
print("Accuracy:", accuracy_score(y_test, predictions))

# Test with new tweet
new_tweet = ["I really love this service"]
new_vec = vectorizer.transform(new_tweet)
result = model.predict(new_vec)

print("Sentiment (1=Positive, 0=Negative):", result[0])
```
