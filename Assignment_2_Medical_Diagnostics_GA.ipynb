AIM: To implement a Genetic Algorithm to optimize feature selection for medical diagnosis (disease prediction).

ALGORITHM:
1. Load medical dataset
2. Split data into features (X) and target (y)
3. Initialize population (random feature subsets)
4. Define fitness function (model accuracy)
5. Perform selection of best individuals
6. Apply crossover (combine parents)
7. Apply mutation (random changes)
8. Repeat for multiple generations
9. Select best feature subset
10. Train final model and predict results

```python
# Import libraries
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import random

# Load dataset (medical dataset)
data = load_breast_cancer()
X = data.data
y = data.target

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Genetic Algorithm parameters
population_size = 10
num_features = X.shape[1]
generations = 5
mutation_rate = 0.1

# Initialize population (random 0/1 for feature selection)
def create_population():
    return [np.random.randint(0, 2, num_features) for _ in range(population_size)]

# Fitness function (accuracy)
def fitness(individual):
    selected_features = [i for i in range(num_features) if individual[i] == 1]

    if len(selected_features) == 0:
        return 0

    model = DecisionTreeClassifier()
    model.fit(X_train[:, selected_features], y_train)
    predictions = model.predict(X_test[:, selected_features])

    return accuracy_score(y_test, predictions)

# Selection
def selection(population):
    population = sorted(population, key=lambda x: fitness(x), reverse=True)
    return population[:2]

# Crossover
def crossover(parent1, parent2):
    point = random.randint(1, num_features - 1)
    child = np.concatenate((parent1[:point], parent2[point:]))
    return child

# Mutation
def mutate(individual):
    for i in range(num_features):
        if random.random() < mutation_rate:
            individual[i] = 1 - individual[i]
    return individual

# Genetic Algorithm process
population = create_population()

for gen in range(generations):
    parents = selection(population)
    new_population = parents.copy()

    while len(new_population) < population_size:
        child = crossover(parents[0], parents[1])
        child = mutate(child)
        new_population.append(child)

    population = new_population

# Best solution
best_individual = max(population, key=lambda x: fitness(x))
best_features = [i for i in range(num_features) if best_individual[i] == 1]

print("Selected Features:", best_features)

# Final model
model = DecisionTreeClassifier()
model.fit(X_train[:, best_features], y_train)
predictions = model.predict(X_test[:, best_features])

print("Final Accuracy:", accuracy_score(y_test, predictions))
```

Genetic Algorithm successfully selected **optimal features and improved classification accuracy.**
