AIM: To implement a Decision Tree algorithm to detect fraudulent credit card transactions.

ALGORITHM:
1. Load dataset (fraud detection data)
2. Preprocess data (handle missing values if any)
3. Split dataset into training and testing
4. Apply Decision Tree classifier
5. Train the model
6. Predict on test data
7. Evaluate accuracy

```python
# Import libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Create simple sample dataset
data = {
    'amount': [100, 2000, 150, 3000, 50, 5000, 120, 7000],
    'location': [1, 0, 1, 0, 1, 0, 1, 0],  # 1 = local, 0 = international
    'is_fraud': [0, 1, 0, 1, 0, 1, 0, 1]
}

df = pd.DataFrame(data)

# Features and target
X = df[['amount', 'location']]
y = df['is_fraud']

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Model
model = DecisionTreeClassifier()

# Train
model.fit(X_train, y_train)

# Predict
predictions = model.predict(X_test)

# Accuracy
print("Accuracy:", accuracy_score(y_test, predictions))

# Test with new data
new_transaction = [[4000, 0]]  # high amount + international
result = model.predict(new_transaction)

print("Fraud Prediction (1=Fraud, 0=Normal):", result[0])
```

The Decision Tree model successfully classified transactions as fraud or normal.
